```python
from simulation.unity_simulator.comm_unity import UnityCommunication
```

```python
comm = UnityCommunication(url="127.0.0.1", port="8080")  # если exe запущен вручную
```

```python
print("connect:", comm.check_connection())

comm.reset(0)
comm.add_character()

script = ["<char0> [walk] <fridge> (1)"]
comm.render_script(script, find_solution=True)

print("OK: script rendered")
```

```python
ok, n = comm.camera_count()
print("camera_count:", ok, n)
```

```python
import matplotlib.pyplot as plt
```

```python
for cam_id in [0, 1, n-1, -1]:
    ok, imgs = comm.camera_image([cam_id], image_width=640, image_height=360)
    print("cam", cam_id, "ok:", ok, "len:", len(imgs))
    if ok and imgs:
        plt.figure()
        plt.imshow(imgs[0])
        plt.axis("off")
        plt.title(f"camera {cam_id}")
        plt.show()
```

```python
from simulation.unity_simulator.comm_unity import UnityCommunication

comm = UnityCommunication(url="127.0.0.1", port="8080")
comm.reset(0)
comm.add_character()

script = ["<char0> [walk] <fridge> (1)"]

success, msg = comm.render_script(
    script=script,
    processing_time_limit=60,
    find_solution=True,
    image_width=640,
    image_height=360,
    skip_animation=False,
    recording=True,
    file_name_prefix="walk_to_fridge"
)
print(success, msg)
```
